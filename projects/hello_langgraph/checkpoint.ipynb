# Checkpointer & Thread

* Thread is similar to chat session with a user
* Every thread will have messgaes history
* By default langgraph does not remeber previous converaion
* to temeber the converation s (state accross nodes ) we need to use checkponting


```python
from langgraph.graph  import StateGraph,START,END
```

```python
from utils import get_llm,draw_grph
```

```python
class MyState(MessageState):
    usr_msg :str
```

```python
llm = get_llm()
def say_hi(state: MyState):
    response = llm.invoke(state['usr_msg'])
    return {"messages": [response]}
```

```python
message_graph = StateGraph(MyState)
message_graph.add_node("llm_hi", say_hi)
message_graph.set_entry_point("llm_hi")
message_graph.set_finish_point("llm_hi")
```

```python
no_mem_graph = message_graph.compile()
no_mem_graph
```

```python
response = no_mem_graph.invoke({'messages': [], 'usr_msg': "What is capital of India?"})
```

```python
response
```

#[refer here ] https://reference.langchain.com/python/langgraph/checkpoints/


```python
from langgraph.checkpoint.memory import InMemorySaver
```

```python
inmem_checkpointer = InMemorySaver()
```

```python
inmem_graph = message_graph.compile(checkpointer=inmem_checkpointer)
```

```python
# configuring thread
from langchain_core.runnables import RunnableConfig
config: RunnableConfig = {"configurable": {"thread_id": "1"}}
```

```python
response = inmem_graph.invoke({'messages': [], 'usr_msg': "What is capital of India?"},config=config)
```
