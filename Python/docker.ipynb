# installing Dcoker in GCP and run n8n command

| **Step** | **Purpose**                                             | **Command**                                                                                                                                                               |
| -------- | ------------------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **1**    | **Update system packages**                              | `sudo apt update && sudo apt upgrade -y`                                                                                                                                  |
| **2**    | **Install Docker**                                      | `sudo apt install docker.io -y`                                                                                                                                           |
| **3**    | **Enable & start Docker**                               | `sudo systemctl enable docker && sudo systemctl start docker`                                                                                                             |
| **4**    | **Check Docker version**                                | `docker --version`                                                                                                                                                        |
| **5**    | **(Optional) Add user to Docker group**                 | `sudo usermod -aG docker $USER && newgrp docker`                                                                                                                          |
| **6**    | **Pull the official n8n Docker image**                  | `docker pull n8nio/n8n`                                                                                                                                                   |
| **7**    | **Run n8n container (simple)**                          | `docker run -d --name n8n -p 5678:5678 n8nio/n8n`                                                                                                                         |
| **8**    | **(Recommended) Run n8n with env + persistent storage** | `bash docker run -d \ --name n8n \ -p 5678:5678 \ -v ~/.n8n:/home/node/.n8n \ -e N8N_HOST=$(curl -s ifconfig.me) \ -e N8N_PORT=5678 \ -e N8N_PROTOCOL=http \ n8nio/n8n `  |
| **9**    | **Check running containers**                            | `docker ps`                                                                                                                                                               |
| **10**   | **Get container logs (to verify startup)**              | `docker logs -f n8n`                                                                                                                                                      |
| **11**   | **Find your VM’s external IP**                          | `curl ifconfig.me` <br>*(or use)*<br>`gcloud compute instances list`                                                                                                      |
| **12**   | **Allow port 5678 in GCP firewall**                     | `bash gcloud compute firewall-rules create allow-n8n \ --allow tcp:5678 \ --source-ranges=0.0.0.0/0 \ --target-tags=n8n-server \ --description="Allow access to n8n UI" ` |
| **13**   | **Tag your VM for the firewall rule**                   | `gcloud compute instances add-tags YOUR_VM_NAME --tags=n8n-server --zone=YOUR_VM_ZONE`                                            ex: gcloud compute instances add-tags instance-20251101-102020 --tags=n8n-server --zone=us-central1-c                                                                                                |
| **14**   | **Access n8n Web URL**                                  | Open in browser → `http://<EXTERNAL_IP>:5678`                                                                                                                             |
